```python
with open("rosalind_ini5_1224_2_dataset.txt", 'r') as inputfile:
    lines = inputfile.readlines()
with open("rosalind_ini5_output.txt", 'w') as outputfile:
    for line in lines[1::2]:
        outputfile.write(line)
```

```python
with open("rosalind_ini5_1224_2_dataset.txt", "r") as inputfile:
    counter = 0
    for line in inputfile:
        counter += 1
        if counter % 2 == 0:
            print(line, end="")
```
