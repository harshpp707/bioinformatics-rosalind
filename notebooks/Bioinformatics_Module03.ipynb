```python
with open("rosalind_ini6_1224_2_dataset.txt", "r") as inputfile:
  words = inputfile.read().split()

word_counts = {}
for word in words:
  if word in word_counts:
      word_counts[word] += 1
  else:
      word_counts[word] = 1

for key, value in word_counts.items():
  print(key, value)
```

## TASK P23.2


```python
with open("rosalind_gc.txt", "r") as inputfile:
    best_id = ""
    best_gc = 0

    for seq_record in parse(inputfile, "fasta"):
        current_gc = gc_fraction(seq_record.seq) * 100
        if current_gc > best_gc:
            best_gc = current_gc
            best_id = seq_record.id

    print(best_id)
    print(best_gc)
```
